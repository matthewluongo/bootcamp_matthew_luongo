## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
big_array = np.arange(1_000_000)
print("Loop: ")
%timeit [x*2 for x in big_array]  # List comprehension

print("Vectorization:")
%timeit big_array * 2  # Vectorized operation
```
